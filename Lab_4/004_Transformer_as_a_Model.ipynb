````python
# helpery z labu 2 (bootstrap_ci, macro_f1) + nowe: paired_bootstrap_diff, token_stats

from sklearn.metrics import f1_score

def macro_f1(y_true, y_pred):
    return f1_score(y_true, y_pred, average="macro")

def bootstrap_ci(y_true, y_pred, metric, n_boot=1000, alpha=0.05, seed=RANDOM_STATE):
    rng = np.random.default_rng(seed)
    y_true, y_pred = np.asarray(y_true), np.asarray(y_pred)
    n = len(y_true)
    stats = []
    for _ in range(n_boot):
        i = rng.integers(0, n, n)
        stats.append(metric(y_true[i], y_pred[i]))
    lo, hi = np.percentile(stats, [100 * alpha / 2, 100 * (1 - alpha / 2)])
    return float(metric(y_true, y_pred)), float(lo), float(hi)
    
def paired_bootstrap_diff(y_true, pred_a, pred_b, metric, n_boot=1000, alpha=0.05, seed=RANDOM_STATE):
    rng = np.random.default_rng(seed)
    y_true, pred_a, pred_b = map(np.asarray, (y_true, pred_a, pred_b))
    n = len(y_true)
    diffs = []
    for _ in range(n_boot):
        i = rng.integers(0, n, n)
        diffs.append(metric(y_true[i], pred_b[i]) - metric(y_true[i], pred_a[i]))
    lo, hi = np.percentile(diffs, [100 * alpha / 2, 100 * (1 - alpha / 2)])
    return float(metric(y_true, pred_b) - metric(y_true, pred_a)), float(lo), float(hi)

def token_stats(tokenizer, texts, max_length):
    texts = list(texts)
    ids = tokenizer(texts, add_special_tokens=True, truncation=False)["input_ids"]
    lens = np.array([len(x) for x in ids])
    n_words = np.array([len(re.findall(r"\w+", t)) for t in texts])
    return {
        "median": float(np.median(lens)),
        "p90": float(np.percentile(lens, 90)),
        "pct_truncated": float((lens > max_length).mean()),
        "tokens_per_word": float(lens.sum() / n_words.sum()),
    }
````

````python
# Proszę przygotować dane i pipeline treningowy.

def label2id(label, classes=CLASSES):
    return sorted(classes).index(label)

def id2label(id, classes=CLASSES):
    return sorted(classes)[id]

for label in CLASSES:
    assert id2label(label2id(label, CLASSES), CLASSES) == label
````

````python
from datasets import Dataset

def prepare_dataset(df, text_col="text_norm", label_col="target"):
    df_prepared = df.loc[:, [text_col, label_col]].copy()
    df_prepared = df_prepared.rename(columns={text_col: "text", label_col:"label"})
    df_prepared[label_col] = df_prepared["label"].map(label2id)
    dataset = Dataset.from_pandas(df_prepared[["text", "label"]], preserve_index=False)
    return dataset

train_ds = prepare_dataset(train_df)
val_ds = prepare_dataset(val_df)

display(train_ds)
display(val_ds)
````

````python
def macro_f1_score(y_true, y_pred):
    return f1_score(y_true, y_pred, average="macro")

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    y_pred = np.argmax(logits, axis=-1)
    return {
        'accuracy': accuracy_score(labels, y_pred),
        'macro_f1': macro_f1_score(labels, y_pred),
        'macro_precision': precision_score(labels, y_pred, average='macro', zero_division=0),
        'macro_recall': recall_score(labels, y_pred, average='macro'),
    }
````

````python
from transformers import DataCollatorWithPadding

MAX_LENGHT = 256
BATCH_SIZE = 16
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.1
EPOCHS = 3
DEVICE = "cpu"
SEED = 42

def tokenize_batch(batch):
    return tokenizer(
        batch["text"],
        truncation=True,
        max_length=MAX_LENGHT  
    )

train_ds_tok = train_ds.map(tokenize_batch, batched=True)
val_ds_tok = val_ds.map(tokenize_batch, batched=True)

data_collator = DataCollatorWithPadding(tokenizer)
from transformers import TrainingArguments

training_args = TrainingArguments(
    output_dir="./model_results",
    num_train_epochs=EPOCHS,
    per_device_train_batch_size=BATCH_SIZE,
    per_device_eval_batch_size=BATCH_SIZE,
    learning_rate=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY,
    eval_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    logging_steps=50,
    seed=SEED
)

trainer = Trainer(
    model=model_for_seq_cls,
    args=training_args,
    train_dataset=train_ds_tok,
    eval_dataset=val_ds_tok,
    processing_class=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics
)

trainer.train()
````
